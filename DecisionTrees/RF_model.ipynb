```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# load the datasets
df = pd.read_csv("RF datasets.csv")
```

```python
# categorial columns conversion to numerical columns using one hot encoding

from sklearn.preprocessing import OneHotEncoder

ohe = OneHotEncoder(sparse_output=False, drop='first')
categorical_cols = ['city_tier', 'department', 'education_level']
ohe_df = pd.DataFrame(
    ohe.fit_transform(df[categorical_cols]), 
    columns=ohe.get_feature_names_out(categorical_cols)
)
df = pd.concat([df.drop(columns=categorical_cols), ohe_df], axis=1)

df.head()
```

```python
x=df.drop(columns=['target_left_company'])
y=df['target_left_company']
```

```python
from sklearn.model_selection import train_test_split

x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier

rf = RandomForestClassifier()
rf.fit(x_train,y_train)
rf_pred = rf.predict(x_test)
```

```python
# compare the actual and predicted values

comparison_df = pd.DataFrame({'Actual': y_test, 'Predicted': rf_pred})
comparison_df.head(10)
```

```python
from sklearn.metrics import classification_report,confusion_matrix,accuracy_score

print("Accuracy Score:",accuracy_score(y_test,rf_pred))
print("Classification Report:\n",classification_report(y_test,rf_pred))
print("Confusion Matrix:\n",confusion_matrix(y_test,rf_pred))
```
