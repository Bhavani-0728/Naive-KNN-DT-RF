Naive Baye's Machine learning Algorithm Implementation

```python
from sklearn.datasets import load_iris
from sklearn.model_selection import train_test_split
from sklearn.naive_bayes import GaussianNB,BernoulliNB,MultinomialNB
from sklearn.metrics import classification_report,accuracy_score,confusion_matrix
```

```python
x,y = load_iris(return_X_y=True)
```

```python
x,y
```

```python
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

```python
gnb = GaussianNB()
gnb.fit(X_train,y_train)
gnb_pred=gnb.predict(X_test) 
print("Accuracy Score using GNB:",accuracy_score(y_test,gnb_pred))
print("Confusion Matrix using GNB:\n",confusion_matrix(y_test,gnb_pred))
print("Classification Report using GNB:\n",classification_report(y_test,gnb_pred))
```

```python
bnb = BernoulliNB()
bnb.fit(X_train,y_train)
bnb_pred=bnb.predict(X_test) 
print("Accuracy Score using BNB:",accuracy_score(y_test,bnb_pred))
print("Confusion Matrix using BNB:\n",confusion_matrix(y_test,bnb_pred))
print("Classification Report using BNB:\n",classification_report(y_test,bnb_pred))
```

```python
mnb = MultinomialNB()
mnb.fit(X_train,y_train)
mnb_pred=mnb.predict(X_test) 
print("Accuracy Score using MNB:",accuracy_score(y_test,mnb_pred))
print("Confusion Matrix using MNB:\n",confusion_matrix(y_test,mnb_pred))
print("Classification Report using MNB:\n",classification_report(y_test,mnb_pred))
```
